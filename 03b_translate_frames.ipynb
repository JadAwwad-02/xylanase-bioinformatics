```python
from mytools import complement, reverse_complement, my_translate
print(my_translate("ATGCGTAAC"))     # MRN
```

```python
def six_frames(seq):
    seq = seq.upper()
    rc = reverse_complement(seq)
    for i in range(3):             
        print("forward", my_translate(seq[i:]))
    for i in range(3):
        print("reverse", my_translate(rc[i:]))
```

```python
six_frames("ATGCGTAACTAG")
```
