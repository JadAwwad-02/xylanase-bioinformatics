```python
def complement(seq):
    seq = seq.upper()
    pairs = { "A" : "T" , "T" : "A" , "G" : "C" , "C" : "G" , "N" : "N" }
    rc = ""
    for base in seq:
        rc = rc + pairs[base]
    return rc
```

```python
dna = "ATGCGTAAC"
complement(dna)
```

```python
dna2 = "aaaatttcccc"
complement(dna2)
```

```python
print(complement("ATGC"))
print(complement("atgcgtaac"))
```

```python
print(complement("ATGC")[::-1])
```
