```python
def reverse_complement(seq):
    seq = seq.upper()
    pairs = { "A" : "T" , "T" : "A" , "G" : "C" , "C" : "G" , "N" : "N" }
    rc = ""
    for base in seq[::-1]:
        rc = rc + pairs[base]
    return rc
```

```python
def is_palindrome(seq):
    seq = seq.upper()
    if seq == reverse_complement(seq):
        return True
    else:
        return False
```

```python
is_palindrome("AAATTTAAATTT")
```

```python
is_palindrome("GAATTC")
```
