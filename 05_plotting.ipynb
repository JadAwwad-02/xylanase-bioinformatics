```python
from mytools import six_frames

gene = open("xyl_full.txt").read().strip()
frames = six_frames(gene)

lengths = []
for f in frames:
    lengths.append(len(f))

print(lengths)
```

```python
labels = ["fwd 0", "fwd 1", "fwd 2", "rev 0", "rev 1", "rev 2"]
```

```python
import matplotlib.pyplot as plt

plt.bar(labels, lengths)
plt.show()
```
