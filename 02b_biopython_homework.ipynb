```python
def find_residue(seq, target):
    positions = []                                   # 4: make empty list once
    for position, aa in enumerate(seq, start=1):     # 4: loop
        if aa == target: 
            positions.append(position)
    return positions
```

```python
W = find_residue(mature, "W" )
Y = find_residue(mature, "Y" )
F = find_residue(mature, "F" )
print(len( W + Y + F ))
```
