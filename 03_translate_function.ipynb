```python
def my_translate(seq):
    seq = seq.upper()
    protein = ""
    for i in range(0, len(seq), 3):
        codon = seq[i:i+3]#
        if codon in table.stop_codons:
            break
        if len(codon) < 3:
            break
        protein = protein + table.forward_table[codon]
    return protein
```

```python
print(my_translate("ATGAGGAAATTTGGGATTTAA"))
```

```python
print(my_translate("ATGCGTAA"))         # 8 letters, not a multiple of 3
```

```python
print(my_translate("ATGCGTAAC"))     # MRN
print(my_translate("atggcctaaggg"))  # MA
print(my_translate("ATGCGTAA"))      # MR, no error
```
