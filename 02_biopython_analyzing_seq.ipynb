```python
def complement(seq):
    seq = seq.upper()
    pairs = {"A": "T", "T": "A", "G": "C", "C": "G", "N": "N"}
    comp = ""
    for base in seq:
        comp = comp + pairs[base]
    return comp

def reverse_complement(seq):
    return complement(seq)[::-1]
```

```python
def find_residue(seq, target):
    positions = []                                   # 4: make empty list once
    for position, aa in enumerate(seq, start=1):     # 4: loop
        if aa == target: 
            positions.append(position)
    return positions
```
