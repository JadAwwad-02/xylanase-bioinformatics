```python
from mytools import complement, reverse_complement, my_translate
print(my_translate("ATGCGTAAC"))
print(complement("ATGCGTAAC"))
print(reverse_complement("ATGCGTAAC"))
```

```python
def six_frames(seq):
    seq = seq.upper()
    rc = reverse_complement(seq)
    frames = []
    for i in range(3):
        frames.append(my_translate(seq[i:]))
    for i in range(3):
        frames.append(my_translate(rc[i:]))
    return frames
```

```python
print(six_frames("ATGCGTAACTAG"))
```

```python
print(len(frames))      # 6
print(frames[0])        # MRN      ← first item
print(frames[3])        # LVTH     ← fourth item
```
